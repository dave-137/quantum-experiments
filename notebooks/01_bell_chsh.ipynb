# Bell/CHSH game

Four-qubit referee/player circuit. The success predicate is `(a ^ b) == (x & y)`. Classical limit 0.75; quantum limit cos(pi/8)^2. This is simulation, not a physical Bell test.

Run after installing the shared environment. Source provenance: [sources](../docs/sources.md); method: [methods](../docs/methods.md). All numerical computation is imported from the study module.


```python
from quantum_experiments.bell_chsh import run
from quantum_experiments.reporting import save_report
from pathlib import Path
import json
binding_path = Path("../proof/binding.json")
binding = json.loads(binding_path.read_text()) if binding_path.exists() else None
result = run(repetitions=4096, seed=137)
report = save_report("bell", result, Path("../results/notebooks/bell"), binding)
{k: result[k] for k in ["wins", "trials", "win_probability", "quantum_limit", "classical_limit"]}
```

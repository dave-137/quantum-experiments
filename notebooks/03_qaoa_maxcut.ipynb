# Six-node Max-Cut QAOA

Cirq donor p=2 Rx/ZZ ansatz and noisy sampled objective optimized with Nelder–Mead. The best sampled cut differs from the mean-cut objective. Exact enumeration supplies the classical optimum; optimizer termination is retained.

Run after installing the shared environment. Source provenance: [sources](../docs/sources.md); method: [methods](../docs/methods.md). All numerical computation is imported from the study module.


```python
from quantum_experiments.qaoa_maxcut import run
from quantum_experiments.reporting import save_report
from pathlib import Path
import json
binding_path = Path("../proof/binding.json")
binding = json.loads(binding_path.read_text()) if binding_path.exists() else None
result = run(repetitions=256, maxiter=50, graph_seed=137, parameter_seed=138, simulator_seed=139)
report = save_report("qaoa", result, Path("../results/notebooks/qaoa"), binding)
{k: result[k] for k in ["graph_edges", "best_sampled_bitstring", "best_sampled_cut", "exact_optimum", "best_cut_ratio", "optimizer"]}
```
